### tkinter란?
- tkinter는 파이썬에서 그래픽 사용자 인터페이스(GUI : graphical user interface)를 개발할 때 필요한 모듈

# 실습

### 버튼 위젯


```python
from tkinter import *       # Tk 모듈 불러오기
window = Tk()               # 창(윈도우) 생성

button = Button(window, text="클릭하세요!",     # 버튼 생성
                bg="yellow", fg="blue",        # 버튼의 색과 배경색 설정
            width = 80, height = 2             # 창(윈도우) 크기 설정
                )

button.pack()           # 버튼 패키징
window.mainloop()       # 사용자가 입력할때까지 대기
```
