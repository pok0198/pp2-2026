### 내장함수와 내장함수의 종류
- 내장함수 : 파이썬(인터프리터)에 포함된 기본 전역 함수



- abs() : 숫자의 절댓값을 반환한다.
- all() : 시퀀스(리스트나 딕셔너리 등)를 받아서, 시퀀스의 모든 항목의 참이면 True를 반환한다.
- any() : 시퀀스 객체에 있는 한 개의 항목이 참인 경우 참
- eval() : 전달된 수식을 구문 분석하고 프로그램 내에서 수식을 실행
- sum() : 리스트에 존재하는 항목을 모두 더하여 합계를 반환
- len() : 객체의 길이를 계산하여 반환
- list(), set(), dict(), tuple() : 각 타입의 비어있는 시퀀스 객체를 생성하여 반환
- map() : 반복가능한 객체(리스트, 튜플 등)의 각 항목에 주어진 함수를 적용한 후, 결과를 반환
- dir() : 객체가 갖고있는 변수나 함수를 보여줌
- max(), min() : max()함수는 리스트나 튜플, 문자열에서 가장 큰 항목을 반환 min() 함수는 그 반대
- enumerate() : 시퀀스를 입력받아 열거형(enumerate) 객체를 반환
- filter() : 특정 조건을 만족하는 요소만을 뽑는다.(말 그대로 필터를 말함)
- zip() : 2개의 시퀀스 자료형을 하나로 묶어주는 함수(2개 이상 가능하다.)
- sorted(시퀀스, key함수, reverse여부) : 시퀀스 객체를 입력받아 원소를 정렬하여 정렬된 시퀀스 객체를 생성하고 반환(key함수는 각 원소의 크기를 반환하는 함수, reverse = Ture이면 내림차순)
- int(), str(), float() : 형변환 함수(가장 기본적인 함수.)
#### 번외
- key 매개변수 : 정렬을 하다 보면 정렬에 사용되는 키를 개발자가 변경해주어야 하는 경우가 종종있다.

```python
class Student:
    def __init__(self, name, grade, number):
        self.name = name
        self.grade = grade
        self.number = number
    def __repr__(self):
        return repr((self.name, self.grade, self.number))

students = [
    Student('홍길동', 3.9, 20160303),
    Student('김철수', 3.0, 20160302),
    Student('최자영', 4.3, 20160301)
    ]
print(sorted(students, key = lambda students : students.number))
```

```python
# map()함수와 람다식

list_a = [1, 2, 3, 4, 5]
f = lambda x : 2*x
result = map(f, list_a)     # map()함수는 반복 가능한(iterable)데이터의 각 요소에 지정된 함수를 일괄적으로 적용하여 새로운 결과(객체 또는 배열)을 반환하는 함수
print(list(result))
```

```python
# filter()함수와 람다식

list_a = [1, 2, 3, 4, 5, 6]
result = filter(lambda x : x % 2 == 0, list_a)      # filter()해당 함수는 말 그대로 조건에 맞춰 필터해서 값을 반환
print(list(result))
```

```python
# reduce()함수와 람다식

import functools
result = functools.reduce(lambda x,y: x+y, [1, 2, 3, 4])
print(result)
```

```python
# 람다식으로 온도 변환 코드
# map() 함수를 사용


# 람다식을 사용하지않은 코드
def celsius(T):
    return (5.0/9.0)*(T-32.0)

f_temp = [0, 10, 20, 30, 40, 50]

c_temp = map(celsius, f_temp)
print(list(c_temp))


# 람다식 사용 코드
f_temp = [0, 10, 20, 30, 40, 50]
c_temp = map(lambda x: (5.0/9.0)*(x-32.0), f_temp)
print(list(c_temp))
```

# 이터레이터 
- 파이썬에는 for 루프와 함께 사용할 수 있는 여러 종류의 객체가 있으며 이들 객체는 이터러블 객체(iterable object)이라고 불린다

#### 객체가 이터러블 객체가 되려면
- __iter__()은 이터러블 객체 자신을 반환
- __next__()은 다음 반복을 위한 값을 반환 만약 더 이상의 값이 없으면 StopIteration 예외를 발생하면 된다.

#### 이터레이터 예제


```python
for i in [1, 2, 3, 4]:
    print( i, end = " ")
```

```python
for c in "python":
    print(c, end = " ")
```

```python
class MyCounter(object):

# 생성자 메소드를 정의
    def __init__(self, low, high):
        self.current = low
        self.high = high

# 이터레이터 객체로서 자신을 반환
    def __iter__(self):
        return self
    def __len__(self):
         return high-low+1
    def __next__(self):

# current가 high 보다 크면 StopIteration 예외를 발생
# current가 high 보다 작으면 다음 값을 반환
        if self.current > self.high:
            raise StopIteration
        else:
            self.current += 1
            return self.current - 1
```

```python
c = MyCounter(1,10)
for i in c:
    print(i, end=' ')
```

### 제너레이터
- 제너레이터(generators)는 키워드 yield를 사용하여서 함수로부터 이터레이터를 생성하는 하나의 방법
    - 정의된 제너레이터를 for루프에서 이터레이터로 사용할 수 있다.

#### 제너레이터 예제

```python

def myGenerater():
    yield 'first'
    yield 'second'
    yield 'third'

for word in myGenerater():
    print(word)
```

```python
def MyCounterGen(low, high):
    while low <= high:
        yield low
        low += 1

for i in MyCounterGen(1, 10):
    print(i, end='  ')
```

#### 피보나치 이터레이터

```python
class FibIterater:
    def __init__(self, a = 1, b = 0, maxValue = 50):
        self.a = a
        self.b = b
        self.maxValue = maxValue

        # __iter__()와 __next__()를 정의하여서 클래스를 이터레이터 클래스로 작성한다.
    def __iter__(self):
        return self

    def __next__(self):
        n = self.a + self.b
        if n > self.maxValue:
            raise StopIteration()       # 파이썬에서 이터레이터(Iterater)가 더 이상 반환할 값이 없을 때 발생하는 예외
        self.a = self.b
        self.b = n
        return n

for i in FibIterater():
    print(i, end=" ")
```

### 이터레이터와 제너레이터의 차이
- 이터레이터는 값을 하나씩 꺼내는 동작을 제공하는 객체이고 제너레이터는 그런 이터레이터를 쉽게 만들 수 있는 기능이다.
    - 다른 방식으로 보면 제너레이터는 이터레이터의 하위 종류라고 생각할 수 있다.

### 연산자 오버로딩
- 연산자를 메소드로 정의하는 것을 연산자 오버로딩(operator overloading)이라고 한다.

- 오버로딩할 수 있는 연산자



#### 오버로딩 예제

```python
class Point:
    def __init__(self, x = 0, y = 0):
        self.x = x
        self.y = y

p1 = Point(1,2)
p2 = Point(3,4)
p1 + p2
```

```python
class Point:
    def __init__(self,x = 0,y = 0):
        self.x = x
        self.y = y
    def __add__(self, other):
        x = self.x + other.x
        y = self.y + other.y
        return Point(x, y)
    def __str__(self):
        return 'Point('+str(self.x)+', '+str(self.y)+')'
p1 = Point(1, 2)
p2 = Point(3, 4)

print(p1+p2)    
```

### 모듈
- 파이썬에서 모듈(module)이란 함수나 변수 또는 클래스들을 모아 놓은 파일이다.
- 파이썬 프로그램이 길어진다면 유지 보수를 쉽게 하기 위해 여러개의 파일로 분할가능 또한 파일을 사용하면 한번 작성한 함수를 복사하지않고 여러 프로그램에서 사용가능
- 프로젝트는 여러 모듈로 구성됨
    - 메인모듈에서 시작하여 여러 모듈을 임포트해서 사용가능
- 폴더로 나누어 그룹화 할수있다.
    - fold/mod.py -> import fold.mod
- 모듈의 일부 함수만 임포트 할 수 있다.
    - from fold.mod import func
- 입력받은 모듈 이름의 별칭을 만들 수 있다.
    - import fold.mod as m
    - m.func()
- 폴더는 라이브러리 목록에 추가하여 검색가능해야 한다.

### 모듈의 종류


- copy 모듈
    - 리스트와 같은 객체를 복사할 때 단수니 리스트를 참조하는 변수만 하나 더 만든다면 실제로 객체가 복사되지않음
    - 객체를 올바르게 복사하려면 copy 모듈의 copy()와 deepcopy() 함수를 사용하여야 한다.
    - copy() 함수는 얕은 복사(shallow copy)를 수행한다. 원본 객체의 최상위 수준의 객체는 복사되지만 내부에 있는 객체들은 원본과 같은 객체를 참조한다.

- deepcopy()
    - deepcopy() 함수는 깊은 복사(deepcopy)를 수행한다. 원본 객체와 그 안에 있는 모든 객체들을새로운 객체로 복사한다.

- sys 모듈
    - sys 모듈은 파이썬 인터프리터에 대한 다양한 정보를 제공하는 모듈이다.

#### 모듈 작성


```python
# 피보나치 수열 모듈
def fib(n):             # 피보나치 수열 출력
    a, b = 0, 1
    while b < n:
        print(b, end='')
        a, b = b, a+b
    print()

def fib2(n):            # 피보나치 수열을 리스트로 반환
    result = []
    a, b = 0, 1
    while b < n:
        result.append(b)
        a, b = b, a+b
    return result
```
