# byte, bytes, bytearray 공부

```python
# byte  한 바이트 크기의 데이터를 다룰때 사용한다.
num = 65
print(num)
print(chr(num))


# bytes 수정할 수 없는 바이트 데이터
data = bytes([65, 66, 67])
print(data)


# bytearray 수정할 수 있는 바이트 데이터
data2 = bytearray([65, 66, 67])
print(data2)

data2[0] = 68
print(data2)

# bytes와 bytearray의 차이는 수정 가능 여부라고 기억하기
```

all(), any() <--함수 공부

```python
# all()
numbers = [True, True, True]
print(all(numbers))

numbers = [True, False, True]
print(all(numbers))


# any()
numbers = [False, False, True]
print(any(numbers))

numbers = [False, False, False]
print(any(numbers))

# all() 모두 True일 때 True 반환, 하나라도 False이면 False 반환
# any() 하나라도 True이면 True 반환, 모두 False이면 False 반환
```
