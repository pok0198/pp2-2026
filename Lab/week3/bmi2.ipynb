# 문제
- 여러 학생들의 키와 몸무게를 리스트로 입력 받아 BMI 리스트를 출력하는 함수와 테스트하는 함수를 작성하시오.
- BMI 함수는 지난 시간에 작성한 get_bmi 함수를 활용하여 작성하시오.

```python
def get_bmi(height, weight):
    return weight / (height / 100) ** 2


def get_bmi_list(heights, weights):
    bmi_list = []

    for i in range(len(heights)):
        bmi = get_bmi(heights[i], weights[i])
        bmi_list.append(round(bmi, 1))

    return bmi_list


def test_get_bmi_list():
    heights = [170, 175, 180]
    weights = [60, 70, 80]

    result = get_bmi_list(heights, weights)

    print(result)


test_get_bmi_list()
```
