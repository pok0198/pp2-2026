```python
# 텍스트 파일 열고 닫기

infile = open("input.txt","r")
infile.close()

# "r" : 읽기 모드 (파일을 처음부터 읽음)
# "w" : 쓰기 모드 (파일을 처음부터 쓴다. 만약 파일이 없다면 생성하고 파일이 있다면 내용은 지워진다.)
# "a" : 추가 모드 (파일의 끝에 내용을 추가한다. 파일이 없으면 생성한다.)
# "r+" : 읽기와 쓰기 모드 (파일을 읽고 쓸 수 있다. 모드를 변경하려면 seek()메소드가 호출되어야 한다.)
```

```python
# 텍스트 파일에서 한 줄 읽기 : '\n'까지 읽기

infile = open("input.txt","r",encoding="utf-8")
# infile = open("input.txt","r") <- 해당 코드로 실행하면 cp949 인코딩 오류가 발생하는데 이는 파일이 utf-8로 인코딩되어 있지만 cp949로 읽으려고 시도하기 때문에 발생하는 오류이다.
line = infile.readline()
while line !="":
    print(line)
    line = infile.readline()
```

```python
# 파일에서 한 줄 읽기 - readline() -> str 반환

infile = open("input.txt","r",encoding="utf-8")
line = infile.readline()
while line != "":
    print(line, end="") # 이유는 모르지만 print(line)로 출력하면 줄바꿈이 두 번 발생한다. 따라서 end=""를 사용해서 줄바꿈을 없애야한다.
    line = infile.readline().strip()
    
    # print(eline)
    # print(infile.tell())
    
    # infile.seek(0)
    # eline = infile.readline().strip()
    # print(eline)
    # print(infile.tell())
```
